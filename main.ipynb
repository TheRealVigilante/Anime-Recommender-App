```python
import numpy as np
import pandas as pd
import tkinter as tk
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
anime_data = pd.read_csv('data/Final.csv')
anime_data.head()
```

```python
anime_data.shape
```

```python
list_of_all_titles = anime_data['Name'].tolist()
```

```python
main_features = ['Score', 'Genres', 'Type', 'Episodes', 'Popularity']
```

```python
combined_features = (
    anime_data['Score'].astype(str) + ' ' +
    anime_data['Genres'].astype(str) + ' ' +
    anime_data['Type'].astype(str) + ' ' +
    anime_data['Episodes'].astype(str) + ' ' +
    anime_data['Popularity'].astype(str)
)

print(combined_features)
```

```python
vectorizer = TfidfVectorizer()
```

```python
feature_vectors = vectorizer.fit_transform(combined_features)
```

```python
print(feature_vectors)
```

```python
cosine_sim = cosine_similarity(feature_vectors)
print(cosine_sim)
print(cosine_sim.shape)
```

```python

def anime_name_input():
    def show_anime_name():
        nonlocal anime_name 
        anime_name = entry.get()
        root.destroy()  

    anime_name = None

    root = tk.Tk()
    root.title("Favorite Anime Input")
    root.geometry("400x200")

    root.configure(bg="#f0f0f0")
    label_font = ("Helvetica", 12)
    entry_font = ("Helvetica", 12)
    button_font = ("Helvetica", 12, "bold")

    label = tk.Label(root, text="Enter the name of your favorite anime:", font=label_font, bg="#f0f0f0")
    label.pack(pady=20)

    entry = tk.Entry(root, width=40, font=entry_font)
    entry.pack(pady=10)

    button = tk.Button(root, text="Submit", command=show_anime_name, font=button_font, bg="#4caf50", fg="white", activebackground="#45a049")
    button.pack(pady=20)

    root.mainloop()

    return anime_name

```

```python
favorite_anime = anime_name_input()
print(f"Favourite anime: {favorite_anime}")
```

```python
print(list_of_all_titles)
```

```python
find_close_match = difflib.get_close_matches(favorite_anime, list_of_all_titles)
print(find_close_match)
```

```python
close_match = find_close_match[0]
print(close_match)
```

```python
index_of_the_movie = anime_data[anime_data.Name == close_match]['Index'].values[0]
print(index_of_the_movie)
```

```python
sim_score = list(enumerate(cosine_sim[index_of_the_movie]))
print(sim_score)
```

```python
len(sim_score)
```

```python
sorted_sim_score = sorted(sim_score, key=lambda x: x[1], reverse=True)
print(sorted_sim_score)
```

```python
from tkinter import ttk
import webbrowser

def open_google_search(query):
    url = f"https://www.google.com/search?q={query}"
    webbrowser.open(url)

def anime_suggestions(anime_data, sorted_sim_score):
    root = tk.Tk()
    root.title("Anime List")
    root.geometry("600x400")

    frame = tk.Frame(root)
    frame.pack(fill=tk.BOTH, expand=True)

    scrollbar = tk.Scrollbar(frame, orient=tk.VERTICAL)
    scrollbar.pack(side=tk.RIGHT, fill=tk.Y)

    canvas = tk.Canvas(frame, yscrollcommand=scrollbar.set)
    canvas.pack(side=tk.LEFT, fill=tk.BOTH, expand=True)

    scrollbar.config(command=canvas.yview)

    canvas_frame = tk.Frame(canvas)
    canvas.create_window((0, 0), window=canvas_frame, anchor=tk.NW)

    i = 1
    for anime in sorted_sim_score:
        if i > 30:  
            break
        index = anime[0]
        title_from_index = anime_data[anime_data.Index == index]['Name'].values[0]
        
        link = tk.Label(canvas_frame, text=f"{i}. {title_from_index}", fg="blue", cursor="hand2")
        link.pack(anchor=tk.W, pady=2)
        link.bind("<Button-1>", lambda e, query=title_from_index: open_google_search(query))
        
        i += 1

    canvas_frame.update_idletasks()
    canvas.config(scrollregion=canvas.bbox("all"))

    root.mainloop()

```

```python
anime_suggestions(anime_data, sorted_sim_score)
```
